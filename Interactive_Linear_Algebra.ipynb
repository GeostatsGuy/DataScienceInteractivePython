The goal of these interactive demonstrations is to connect the fundamental operations of linear algebra to their applications in machine learning and neural network architectures.

Students first learn the mathematical operations, then explore how these same operations are used to process, combine, transform, and selectively control information in neural networks.

Each demonstration includes:

1. A familiar linear algebra operation.
2. A connection to a machine learning or neural network application.
3. A custom Matplotlib visualization.
4. Interactive ipywidgets sliders to change inputs and immediately observe the consequences.
5. Explicit mathematical equations and numerical examples.

**Let's not just show the math. Let's show what the math does!**

#### Recommended Teaching Sequence

| Order | Linear Algebra | Neural Network Connection | Main Insight |
|---|---|---|---|
| 1 | Vector addition | LSTM memory / residual connections | Combine information |
| 2 | Dot product | Artificial neuron | Weighted combination |
| 3 | Matrix-vector multiplication | Dense ANN layer | Multiple neurons simultaneously |
| 4 | Element-wise multiplication | Gates / attention-related features | Selectively control information |

#### Common Visualization Design

To create a consistent learning experience across all four demonstrations:

- Use 4–6 elements in the vectors, allowing students to explore multidimensional operations.
- Use consistent vector colors across all demonstrations.
- Display equations immediately above the visualization.
- Show numerical values alongside graphical representations.
- Make widgets respond immediately, without requiring a Run button.
- Start with simple, deliberately selected numerical values, then allow students to experiment.

#### Student Exploration Challenges

Include a short *What happens if...?* challenge in each demonstration.

Examples:

1. **LSTM memory:** What happens to the memory state when the forget gate approaches zero?
2. **Artificial neuron:** What happens when one weight becomes negative?
3. **Dense layer:** What happens when one row of the weight matrix is multiplied by zero?
4. **Feature gating:** What happens when all gate values approach zero?

Students manipulate the controls and discover the answers themselves.

#### Overall Learning Objective

Neural networks are not magic. They are built from familiar linear algebra operations, applied repeatedly, combined with nonlinear activation functions, and optimized through learning.

**Learn the operation. Explore the visualization. Discover the neural network connection!**

#### Declare Plotting Functions

Let's declare 2 convenience functions for plotting our results,

* **add_arrow** - formatted arrow for flow charts / deep learning architectures

* **add_box** - formatted boxes for flow charts / deep learning architectures

```python
def add_arrow(ax, start, end,color='black',lw=1.8):           # function to create arrows
    arrow = FancyArrowPatch( start, end, arrowstyle='->', mutation_scale=15, linewidth=lw, color=color)
    ax.add_patch(arrow)

def add_box(ax, x, y, width, height, label, color):           # function to create rounded boxes
    box = FancyBboxPatch((x, y), width, height,boxstyle="round,pad=0.02",facecolor=color,
            edgecolor='black',linewidth=1.5,alpha=0.85)
    ax.add_patch(box)
    ax.text(x + width/2, y + height/2,label, ha='center', va='center',fontsize=10, weight='bold')
```

#### 1. Vector Addition and Element-wise Multiplication – Combining Information for the LSTM Memory Update

**Linear algebra operation:** Vector addition

**Machine learning connection:** Long Short-Term Memory (LSTM) networks

Demonstrate how an LSTM updates its long-term memory by combining retained historical information with new incoming information.

The LSTM cell-state update equation is:

$$
\bf c_t = \bf f_t \odot \bf c_{t-1} + \bf i_t \odot \bf g_t
$$

Where:

- $\bf c_{t-1}$: previous long-term memory.
- $\bf f_t$: forget gate (how much previous memory to retain).
- $\bf g_t$: candidate new information.
- $\bf i_t$: input gate (how much new information to accept).
- $\odot$: vector element-wise multiplication.
- $+$: vector addition

An LSTM balances retaining historical information with,

* **vector element-wise multiplication** - what information to keep and what information to pass

* **vector addition** - integrating the new information with vector element-size 


```python
c_previous = np.array([1.0, 2.0])                             # previous memory
g_new = np.array([3.0, 1.0])                                  # Candidate new information

def plot_lstm_memory(forget_gate=0.8, input_gate=0.5):        # interactive visualization function
    c_retained = forget_gate * c_previous                     # calculate retained memory and gated new information
    g_gated = input_gate * g_new
    c_updated = c_retained + g_gated                          # updated long-term memory

    fig, (ax, ax2) = plt.subplots(1, 2, figsize=(15, 7))      # create figure

    fig.subplots_adjust(wspace=0.25)

    ax.quiver(0, 0, c_retained[0], c_retained[1],angles='xy', scale_units='xy', scale=1,color='royalblue', # retained memory vector
              width=0.012,label='Retained memory')     
    ax.quiver(c_retained[0], c_retained[1], g_gated[0], g_gated[1], angles='xy', scale_units='xy', scale=1, # gated new information vector 
              color='darkorange', width=0.012, label='Gated new information')
    ax.quiver(0, 0, c_updated[0], c_updated[1], angles='xy', scale_units='xy', scale=1, color='forestgreen', width=0.012, # updated memory vector
              label='Updated memory')
    ax.plot([c_retained[0], c_updated[0]], [c_retained[1], c_updated[1]], 'k--', alpha=0.5) # dashed line connecting vector endpoints
    ax.scatter([0, c_retained[0], c_updated[0]],[0, c_retained[1], c_updated[1]], color='black', s=45, zorder=5) # mark vector endpoints

    ax.text(c_retained[0]-0.25, c_retained[1]+0.15,'Retained', color='royalblue') # labels
    ax.text(c_updated[0]+0.05, c_updated[1]+0.05,'Updated', color='forestgreen')

    results = (                                               # display numerical results
        f"Retained memory: [{c_retained[0]:.2f}, {c_retained[1]:.2f}]\n"
        f"Gated new information: [{g_gated[0]:.2f}, {g_gated[1]:.2f}]\n"
        f"Updated memory: [{c_updated[0]:.2f}, {c_updated[1]:.2f}]"
    )
    ax.text(0.02, 0.98, results, transform=ax.transAxes,verticalalignment='top', fontsize=10,bbox=dict(boxstyle='round', facecolor='white',
        alpha=0.9))


    q0 = ax.quiver(0, 0, c_previous[0], c_previous[1],angles='xy', scale_units='xy', scale=1, # all previous memory
         color='royalblue', alpha=0.5,width=0.008,label='Original memory')

    q00 = ax.quiver(c_previous[0], c_previous[1],g_new[0], g_new[1],angles='xy', scale_units='xy', scale=1, # all new information
                color='darkorange', alpha=0.5, width=0.008, label='Original new information')

    ax.set_xlim(-0.5, 4.5); ax.set_ylim(-0.5, 4.5); ax.set_aspect('equal'); ax.set_xlabel('Memory Component 1') # formatting
    ax.set_ylabel('Memory Component 2'); ax.set_title('LSTM Memory Update: Vector Addition and Element-wise Multiplication')
    ax.grid(True, linestyle='--', alpha=0.4); ax.legend(loc='lower right')
    ax2.set_xlim(0, 1); ax2.set_ylim(0, 1); ax2.axis('off')   # right panel: LSTM memory update diagram
    ax2.set_title('LSTM Memory Update Architecture',fontsize=14, pad=15)
        
    add_box(ax2, 0.03, 0.78, 0.23, 0.10, 'Long-term\nMemory','white') # input boxes
    
    add_box(ax2, 0.03, 0.56, 0.23, 0.10, 'Forget Gate', 'white')
    
    add_box(ax2, 0.03, 0.30, 0.23, 0.10, 'New Information', 'white')
    
    add_box(ax2, 0.03, 0.08, 0.23, 0.10, 'Input Gate', 'white')

    ax2.add_patch(Circle((0.43, 0.73), 0.045, facecolor='white', edgecolor='black', linewidth=2)) # multiplication nodes
    
    ax2.text(0.43, 0.73, '×',ha='center', va='center', fontsize=17)
    
    ax2.add_patch(Circle((0.43, 0.25), 0.045,facecolor='white',edgecolor='black',linewidth=2))
    
    ax2.text(0.43, 0.25, '×', ha='center', va='center', fontsize=17)
    
    ax2.add_patch(Circle((0.70, 0.49), 0.045,facecolor='white',edgecolor='black',linewidth=2)) # addition node
    
    ax2.text(0.70, 0.49, '+',ha='center', va='center', fontsize=17)
    add_box(ax2, 0.81, 0.43, 0.17, 0.12,'Updated\nLong-term\nMemory','white') # updated memory output
    
    add_arrow(ax2, (0.28, 0.83), (0.385, 0.75))               # connect input boxes to gates
    add_arrow(ax2, (0.28, 0.61), (0.385, 0.72))
    
    add_arrow(ax2, (0.28, 0.35), (0.385, 0.27))
    add_arrow(ax2, (0.28, 0.13), (0.385, 0.23))
    
    add_arrow(ax2, (0.475, 0.73), (0.66, 0.52),color='royalblue',lw=2.5) # connect gates to addition node
    add_arrow(ax2, (0.475, 0.25), (0.66, 0.46),color='darkorange',lw=2.5)
    add_arrow(ax2, (0.745, 0.49), (0.80, 0.49))               # connect addition to updated memory
        
    ax2.text(0.43, 0.83, 'Retain',ha='center', fontsize=9, color='black')
    ax2.text(0.55, 0.79, 'Element-wise Martix Multiplication',ha='center', fontsize=9, color='black')
    
    ax2.text(0.43, 0.14, 'Incorporate',ha='center', fontsize=9, color='black')
    ax2.text(0.55, 0.10, 'Element-wise Martix Multiplication',ha='center', fontsize=9, color='black')
    
    ax2.text(0.70, 0.38, 'Combine',ha='center', fontsize=9, color='black')
    ax2.text(0.70, 0.34, 'Matrix Addition',ha='center', fontsize=9, color='black')
    
    plt.show()
    
l = widgets.Text(value='                           Neural Network Linear Algebra Demonstration, Prof. Michael Pyrcz, The University of Texas at Austin',layout=Layout(width='950px', height='30px'))

# Create interactive sliders
forget_slider = widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.05,description='Forget Gate:',continuous_update=False,
    style={'description_width': 'initial'})

input_slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, step=0.05,description='Input Gate:',continuous_update=False,
    style={'description_width': 'initial'})

dashboard = widgets.interactive_output(plot_lstm_memory,{'forget_gate': forget_slider,'input_gate': input_slider}) # link sliders 

slider_box = widgets.HBox([forget_slider, input_slider],layout=widgets.Layout(justify_content='center', gap='30px')) # stack sliders

display(widgets.VBox([l, slider_box, dashboard ]))            # display complete dashboard
```

## 2. Dot Product – A Single Artificial Neuron

**Linear algebra operation:** Dot product

**Machine learning connection:** Artificial neural networks

Demonstrate how an artificial neuron combines multiple inputs using learned weights and a bias, then applies a nonlinear activation function.

$$
z = \bf w \cdot \bf x + b
$$

$$
y = \sigma(z)
$$

Where:

- $\bf x$: input feature vector.
- $\bf w$: learned weight vector.
- $b$: bias.
- $z$: weighted sum before activation.
- $\sigma$: activation function.
- $y$: neuron output.

and ReLU activation, 

$$
y = \text{max} \left( 0, z \right)
$$

**Interactive controls:**

- Three input values ($x_1,x_2$).
- Three weights ($w_1,w_2$).


A neuron combines information from all preivous nodes in the network,

* **dot product** - the node combines all weighted connection as a vector dot product.

```python
def plot_neuron(x1, x2, w1, w2):
    b = 0.1                                                   # define bias
    c1 = w1 * x1                                              # calculate weighted contributions
    c2 = w2 * x2
    z = c1 + c2 + b                                           # calculate neuron output
    y = max(0, z)                                             # ReLU activation

    fig, ax = plt.subplots(1, 2, figsize=(11, 4))             # create figure
    plt.close(fig)

    contributions = [c1, c2, b]                               # left weight contributions
    labels = ['Input 1', 'Input 2', 'Bias']

    bars = ax[0].bar(labels, contributions, color=['steelblue', 'orange', 'gray'])

    ax[0].axhline(0, color='black', linewidth=0.8); ax[0].set_ylabel('Contribution')
    ax[0].set_title('Weighted Contributions'); ax[0].grid(axis='y', alpha=0.3)

    for bar, value in zip(bars, contributions):
        ax[0].text(bar.get_x() + bar.get_width()/2,
                   value + (0.025 if value >= 0 else -0.045),
                   f'{value:.2f}', ha='center',
                   va='bottom' if value >= 0 else 'top')
    ax[0].set_ylim([-1,1])

    ax[1].axis('off'); ax[1].set_xlim(0, 1); ax[1].set_ylim(0, 1) # node schematic
    ax[1].text(0.05, 0.72, f'x1 = {x1:.2f}', ha='center')     # input labels
    ax[1].text(0.05, 0.28, f'x2 = {x2:.2f}', ha='center')

    neuron = plt.Circle((0.55, 0.50), 0.13, fill=False, linewidth=2) # neuron
    ax[1].add_patch(neuron)
    ax[1].text(0.55, 0.46, 'pre-ReLU', ha='center', va='center')
    ax[1].text(0.55, 0.51, f'z = {z:.2f}', ha='center', va='center') # preactivated
    
    ax[1].annotate('', xy=(0.42, 0.57), xytext=(0.17, 0.72),arrowprops=dict(arrowstyle='->', lw=1.8,color='royalblue')) # arrows and weights
    ax[1].annotate('', xy=(0.42, 0.43), xytext=(0.17, 0.28),arrowprops=dict(arrowstyle='->', lw=1.8,color='darkorange'))

    ax[1].text(0.28, 0.72, f'$\lambda_1$ = {w1:.1f}', ha='center'); ax[1].text(0.28, 0.24, f'$\lambda_2$ = {w2:.1f}', ha='center')
    ax[1].text(0.55, 0.30, f'b = {b:.2f}', ha='center')       # bias

    ax[1].annotate('', xy=(0.92, 0.50), xytext=(0.68, 0.50),arrowprops=dict(arrowstyle='->', lw=1.8)) # output
    ax[1].text(0.82, 0.58, f'y = {y:.2f}', ha='center')
    ax[1].set_title(f'Neuron: z = {z:.2f}, y = {y:.2f}')

    neuron = plt.Circle((0.55, 0.50), 0.13, fill=False, linewidth=2)
    ax[1].add_patch(neuron)
    plt.tight_layout()
    display(fig); plt.close(fig)

# Demonstration title
l = widgets.Text(
    value='                           Neural Network Linear Algebra Demonstration - Dot Product, Prof. Michael Pyrcz, The University of Texas at Austin',
    layout=Layout(width='950px', height='30px')
)

# Create interactive sliders
x1_slider = widgets.FloatSlider(value=0.6, min=0.0, max=1.0, step=0.05,description='Input 1:',continuous_update=False,
    style={'description_width': 'initial'})

x2_slider = widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.05,description='Input 2:',continuous_update=False,
    style={'description_width': 'initial'})

w1_slider = widgets.FloatSlider(value=0.8, min=-1.0, max=1.0, step=0.05,description=r'$\lambda_1$:',continuous_update=False,
    style={'description_width': 'initial'})

w2_slider = widgets.FloatSlider(value=-0.4, min=-1.0, max=1.0, step=0.05,description=r'$\lambda_2$:',continuous_update=False,
    style={'description_width': 'initial'})

dashboard = widgets.interactive_output(plot_neuron,{'x1': x1_slider, 'x2': x2_slider, 'w1': w1_slider, 'w2': w2_slider}) # link sliders 

slider_box = widgets.HBox([x1_slider, x2_slider,w1_slider,w2_slider],layout=widgets.Layout(justify_content='center', gap='30px')) # stack sliders

display(widgets.VBox([l, slider_box, dashboard])) # display complete dashboard
```

## 3. Matrix-Vector Multiplication – A Fully Connected Neural Network Layer

**Linear algebra operation:** Matrix-vector multiplication

**Machine learning connection:** Dense neural network layers,

* 3 hidden layer nodes $X_1, X_2, X_3$

* 3 output nodes, pre-activation, $Z_1, Z_2, Z_3$

* 3 output nodes, post-activation, $Y_1, Y_2, Y_3$

Demonstrate how a single matrix-vector multiplication performs multiple weighted sums simultaneously, representing multiple neurons in a fully connected neural network layer.

$$
\bf y = A\bf x + \bf b
$$

For example, a three-input, three-neuron layer:

$$
\begin{bmatrix}
y_1\\y_2\\y_3
\end{bmatrix}
=
\begin{bmatrix}
w_{11}&w_{12}&w_{13}\\
w_{21}&w_{22}&w_{23}\\
w_{31}&w_{32}&w_{33}
\end{bmatrix}
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
+
\begin{bmatrix}
b_1\\b_2\\b_3
\end{bmatrix}
$$

and ReLU activation, 

$$
\begin{bmatrix}
y_1\\y_2\\y_3
\end{bmatrix}
=
\max\left(0,
\begin{bmatrix}
z_1\\z_2\\z_3
\end{bmatrix}
\right)
$$

We assume these connection weights and biases,

$$
\begin{bmatrix}
z_1\\z_2\\z_3
\end{bmatrix}
=
\begin{bmatrix}
0.7&-0.4&0.2\\
-0.3&0.8&0.5\\
0.6&0.1&-0.7
\end{bmatrix}
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
+
\begin{bmatrix}
0.1\\0.2\\0.1
\end{bmatrix}
$$

**Interactive controls:**

- Change input vector values.
- Change individual matrix coefficients.
- Optionally select individual rows of the matrix.

**Custom visualization:**

- Heatmap of the weight matrix.
- Input vector.
- Highlighted row-by-row multiplication.
- Output vector with changing bar heights.

**Teaching message:** Matrix-vector multiplication performs multiple weighted sums simultaneously. This is the foundation of a dense neural network layer.

```python
def plot_dense_layer(x1, x2, x3):
    x = np.array([x1, x2, x3])                                # define input vector
    W = np.array([[ 0.7, -0.4,  0.2],                         # define weight matrix and bias vector
                  [-0.3,  0.8,  0.5],
                  [ 0.6,  0.1, -0.7]])

    b = np.array([0.1, 0.2, 0.1])
    z = np.dot(W, x) + b                                      # matrix-vector multiplication
    y = np.maximum(0, z)                                      # ReLU activation

    fig, ax = plt.subplots(1, 3, figsize=(13, 4))             # create figure

    ax[0].bar(['x1', 'x2', 'x3'], x, color='steelblue')       # plot 1: input vector
    ax[0].set_ylim(0, 1.1); ax[0].set_title('Input Vector'); ax[0].set_ylabel('Input Value'); ax[0].grid(axis='y', alpha=0.3)

    for i, value in enumerate(x):
        ax[0].text(i, value + 0.025, f'{value:.2f}', ha='center')

    ax[1].bar(['z1', 'z2', 'z3'], z,color=['steelblue' if v >= 0 else 'orange' for v in z]) # plot 2: pre-activation values
    ax[1].set_ylim(-1, 2.0); ax[1].axhline(0, color='black', linewidth=0.8); ax[1].set_title('Pre-activation: W x + b')
    ax[1].set_ylabel('Pre-activation'); ax[1].grid(axis='y', alpha=0.3)

    for i, value in enumerate(z):
        ax[1].text(i, value + (0.025 if value >= 0 else -0.045),f'{value:.2f}', ha='center',va='bottom' if value >= 0 else 'top')

    ax[2].bar(['y1', 'y2', 'y3'], y, color='seagreen'); ax[2].set_title('ReLU Activated Output')  # plot 3: activated outputs
    ax[2].set_ylim(0, 1.1); ax[2].set_ylabel('Output Value'); ax[2].grid(axis='y', alpha=0.3)

    for i, value in enumerate(y):
        ax[2].text(i, value + 0.025, f'{value:.2f}', ha='center')

    plt.tight_layout(); display(fig); plt.close(fig)

l = widgets.Text(                                             # demonstration title
    value='                           Neural Network Linear Algebra Demonstration, Prof. Michael Pyrcz, The University of Texas at Austin',
    layout=Layout(width='950px', height='30px'))

x1_slider = widgets.FloatSlider(value=0.6, min=0.0, max=1.0, step=0.05,description='Input 1:', # create interactive sliders
    continuous_update=False,style={'description_width': 'initial'})

x2_slider = widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.05,description='Input 2:',
    continuous_update=False,style={'description_width': 'initial'})

x3_slider = widgets.FloatSlider(value=0.4, min=0.0, max=1.0, step=0.05,description='Input 3:',
    continuous_update=False,style={'description_width': 'initial'})

dashboard = widgets.interactive_output(plot_dense_layer,{'x1': x1_slider, 'x2': x2_slider, 'x3': x3_slider}) # link sliders 

slider_box = widgets.HBox([x1_slider, x2_slider, x3_slider],layout=widgets.Layout(justify_content='center', gap='25px')) # stack sliders
 
display(widgets.VBox([l, slider_box, dashboard]))             # display complete dashboard
```

## 4. Element-wise Multiplication – Attention and Feature Gating

**Linear algebra operation:** Element-wise multiplication

**Machine learning connection:** Neural network gates and attention-related feature processing

Demonstrate how a neural network can selectively control the information passed through its architecture.

For a simple feature-gating example with element-wise multiplication:

$$
\bf y = \bf g \odot \bf x
$$

Where:

- $\bf x$: incoming feature vector.
- $\bf g$: gate values.
- $\bf y$: filtered feature vector.
- $\odot$ indicates element-wise multiplication

**Interactive controls:**

- Input feature values.
- Gate values (0 to 1).
- Slider to adjust overall gate strength.

Use color intensity or opacity to show which features are being suppressed or retained.

Neural networks don't just combine information; they selectively control how much information passes through.

This provides a natural bridge to attention mechanisms, although Transformer attention involves additional operations, including query-key similarities, softmax normalization, and weighted combinations of value vectors.

```python
def plot_feature_gating(x1, x2, x3, g1, g2, g3, gate_strength):

    x = np.array([x1, x2, x3])                                # define input feature vector
    g = np.array([g1, g2, g3])                                # define gate vector
    g_effective = gate_strength * g                           # apply overall gate strength
    y = g_effective * x                                       # element-wise multiplication

    fig, ax = plt.subplots(1, 3, figsize=(13, 4))             # create figure

    labels = ['Feature 1', 'Feature 2', 'Feature 3']
    positions = np.arange(3)

    ax[0].bar(positions, x, color='steelblue'); ax[0].set_ylim(0, 1.1) # plot 1: original input features
    ax[0].set_xticks(positions); ax[0].set_xticklabels(labels); ax[0].set_title('Original Features')
    ax[0].set_ylabel('Feature Value'); ax[0].grid(axis='y', alpha=0.3)

    for i, value in enumerate(x):
        ax[0].text(i, value + 0.025, f'{value:.2f}', ha='center')

    gate_colors = [plt.cm.Blues(0.25 + 0.7 * value)           # plot 2: effective gate values
                   for value in g_effective]

    ax[1].bar(positions, g_effective, color=gate_colors); ax[1].set_xticks(positions); ax[1].set_xticklabels(labels)
    ax[1].set_ylim(0, 1.1); ax[1].set_title('Effective Gate Values'); ax[1].set_ylabel('Gate Strength')
    ax[1].grid(axis='y', alpha=0.3)

    for i, value in enumerate(g_effective):
        ax[1].text(i, value + 0.025, f'{value:.2f}', ha='center')

    ax[2].bar(positions, y, color=gate_colors)                # plot 3: filtered output features

    ax[2].set_xticks(positions); ax[2].set_xticklabels(labels); ax[2].set_ylim(0, 1.1)
    ax[2].set_title('Filtered Features'); ax[2].set_ylabel('Output Value'); ax[2].grid(axis='y', alpha=0.3)

    for i, value in enumerate(y):
        ax[2].text(i, value + 0.025, f'{value:.2f}', ha='center')

    fig.suptitle(f'Element-wise Feature Gating | Overall Gate Strength = {gate_strength:.2f}') # overall title

    plt.tight_layout(); display(fig); plt.close(fig)


l = widgets.Text(
    value='                           Neural Network Linear Algebra Demonstration, Prof. Michael Pyrcz, The University of Texas at Austin',
    layout=Layout(width='950px', height='30px'))

x1_slider = widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.05,description='Feature 1:', # input feature sliders
    continuous_update=False,style={'description_width': 'initial'})

x2_slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, step=0.05,description='Feature 2:',
    continuous_update=False,style={'description_width': 'initial'})

x3_slider = widgets.FloatSlider(value=0.9, min=0.0, max=1.0, step=0.05,description='Feature 3:',
    continuous_update=False,style={'description_width': 'initial'})

g1_slider = widgets.FloatSlider(value=0.9, min=0.0, max=1.0, step=0.05,description='Gate 1:', # individual gate sliders
    continuous_update=False,style={'description_width': 'initial'})

g2_slider = widgets.FloatSlider(value=0.3, min=0.0, max=1.0, step=0.05,description='Gate 2:',
    continuous_update=False,style={'description_width': 'initial'})

g3_slider = widgets.FloatSlider(value=0.6, min=0.0, max=1.0, step=0.05,description='Gate 3:',
    continuous_update=False,style={'description_width': 'initial'})

strength_slider = widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.05,description='Overall Gate Strength:', # overall gate strength slider
    continuous_update=False,style={'description_width': 'initial'})

dashboard = widgets.interactive_output(                       # link sliders to plotting function
    plot_feature_gating,{'x1': x1_slider,'x2': x2_slider,'x3': x3_slider,'g1': g1_slider,'g2': g2_slider,
        'g3': g3_slider,'gate_strength': strength_slider})

input_box = widgets.HBox([x1_slider, x2_slider, x3_slider],layout=widgets.Layout(justify_content='center', gap='25px')) # stack sliders
gate_box = widgets.HBox([g1_slider, g2_slider, g3_slider],layout=widgets.Layout(justify_content='center', gap='25px'))
strength_box = widgets.HBox([strength_slider],layout=widgets.Layout(justify_content='center')) # display overall gate strength

display(widgets.VBox([l,input_box,gate_box,strength_box,dashboard])) # display complete dashboard
```
